```python
import os
import zipfile
import yaml
from pathlib import Path
from ultralytics import YOLO

# 1. Define paths
zip_path = "/content/Bike helmet.v1i.yolo26.zip"
extract_path = "/content/helmet_dataset"
yaml_path = Path(extract_path) / "data.yaml"
model_path = "/content/runs/detect/helmet_workshop/weights/best.pt"
video_path = "/content/test1.mp4"

# 2. Extract dataset and configure paths if not already present
if not os.path.exists(extract_path):
    assert os.path.exists(zip_path), f"ZIP not found at {zip_path}. Please re-upload it first."
    print("Extracting dataset...")
    with zipfile.ZipFile(zip_path, "r") as zip_ref:
        zip_ref.extractall(extract_path)

    if yaml_path.exists():
        config = yaml.safe_load(yaml_path.read_text())
        config["path"] = extract_path
        with open(yaml_path, "w") as f:
            yaml.dump(config, f)
        print("Dataset extracted and data.yaml configured.")

# 3. Train YOLO to regenerate best.pt
print("Training YOLO model...")
model = YOLO("yolov8n.pt")
model.train(
    data=str(yaml_path),
    epochs=30,
    imgsz=640,
    project="/content/runs/detect",
    name="helmet_workshop"
)

# 4. Run inference on the video
assert os.path.exists(video_path), f"Video not found at {video_path}. Please upload test1.mp4."
print("Starting video detection...")
trained_model = YOLO(model_path)
results = trained_model.predict(
    source=video_path,
    conf=0.25,
    device=0,
    save=True,
    project="/content/runs/detect",
    name="video_output",
    exist_ok=True
)

print("Processing complete! Output saved under /content/runs/detect/video_output")
```

```python
from ultralytics import YOLO
from IPython.display import display, Image
from pathlib import Path

# Load your trained helmet detector
MODEL_PATH = "/content/runs/detect/helmet_workshop/weights/best.pt"
IMAGE_PATH = "/content/del.jpeg"
OUTPUT_DIR = "/content/helmet_predictions"

assert Path(MODEL_PATH).exists(), f"Model not found: {MODEL_PATH}"
assert Path(IMAGE_PATH).exists(), f"Image not found: {IMAGE_PATH}"

model = YOLO(MODEL_PATH)

# Run inference
results = model.predict(
    source=IMAGE_PATH,
    imgsz=640,
    conf=0.25,
    device=0,
    save=True,
    project=OUTPUT_DIR,
    name="result",
    exist_ok=True
)

# Display the exact output file returned by YOLO
annotated_path = Path(results[0].save_dir) / Path(IMAGE_PATH).name

print("Input image:", IMAGE_PATH)
print("Annotated image:", annotated_path)
print("Detections:", len(results[0].boxes))

if annotated_path.exists():
    display(Image(filename=str(annotated_path)))
else:
    print("Output file not found:", annotated_path)
```

```python
from ultralytics import YOLO
from google.colab import files
from pathlib import Path
import os

model_path = "/content/runs/detect/helmet_workshop/weights/best.pt"
video_path = "/content/test1.mp4"

if os.path.exists(model_path) and os.path.exists(video_path):
    print("Starting video detection (filtering for helmets only)...")
    trained_model = YOLO(model_path)

    # Run prediction, looking ONLY for class 0 (with_helmet)
    results = trained_model.predict(
        source=video_path,
        conf=0.35,
        classes=[0],        # Ignores 'without_helmet' completely
        device=0,
        save=True,
        project="/content/runs/detect",
        name="video_output_only_helmets",
        exist_ok=True
    )

    # Automatically find and download the new video
    out_dir = Path("/content/runs/detect/video_output_only_helmets")
    video_files = list(out_dir.glob("*.avi")) + list(out_dir.glob("*.mp4"))

    if video_files:
        print(f"Downloading: {video_files[0].name}")
        files.download(str(video_files[0]))
else:
    print("Error: Missing model or video file. Check your uploads!")
```
